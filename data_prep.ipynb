```python
import pandas as pd
series_df = pd.read_csv('./downloaded_data/series_data.txt', delimiter='\t')
state_df = pd.read_csv('./downloaded_data/state_level_data.tsv', dtype={'       value': str},  delimiter="\t")
# Keep only years >= 2019 and less than 2025
state_df = state_df[(state_df['year'] >= 2019) & (state_df['year'] <= 2023)]

# Remove the year average
state_df = state_df[state_df['period'] != 'M13']



state_series_join = pd.merge(state_df, series_df, on= 'series_id                     ')

state_series_join['Employment Measure'] = state_series_join['series_title'].str.split(':').str[0]
state_series_join['State'] = state_series_join['series_title'].str.split(':').str[1].str.split('(').str[0].str.strip()

state_abbreviations = {
    'Alabama': 'AL', 'Alaska': 'AK', 'Arizona': 'AZ', 'Arkansas': 'AR',
    'California': 'CA', 'Colorado': 'CO', 'Connecticut': 'CT',
    'Delaware': 'DE', 'Florida': 'FL', 'Georgia': 'GA', 'Hawaii': 'HI',
    'Idaho': 'ID', 'Illinois': 'IL', 'Indiana': 'IN', 'Iowa': 'IA',
    'Kansas': 'KS', 'Kentucky': 'KY', 'Louisiana': 'LA', 'Maine': 'ME',
    'Maryland': 'MD', 'Massachusetts': 'MA', 'Michigan': 'MI',
    'Minnesota': 'MN', 'Mississippi': 'MS', 'Missouri': 'MO',
    'Montana': 'MT', 'Nebraska': 'NE', 'Nevada': 'NV', 'New Hampshire': 'NH',
    'New Jersey': 'NJ', 'New Mexico': 'NM', 'New York': 'NY',
    'North Carolina': 'NC', 'North Dakota': 'ND', 'Ohio': 'OH',
    'Oklahoma': 'OK', 'Oregon': 'OR', 'Pennsylvania': 'PA',
    'Puerto Rico': 'PR', 
    'Rhode Island': 'RI', 'South Carolina': 'SC', 'South Dakota': 'SD',
    'Tennessee': 'TN', 'Texas': 'TX', 'Utah': 'UT', 'Vermont': 'VT',
    'Virginia': 'VA', 'Washington': 'WA', 'West Virginia': 'WV',
    'Wisconsin': 'WI', 'Wyoming': 'WY'
}



# swap keys and values in the dictionary
state_lookup = dict((v, k) for k, v in state_abbreviations.items())
state_series_join['State Abbreviation'] = state_series_join['State'].map(state_abbreviations)

state_series_join.to_csv('state_series_join.csv', index=False)


county_df = pd.read_csv('./downloaded_data/county_data.tsv', dtype={'       value': str}, delimiter='\t')
# Keep only years >= 2019 & <= 2023
county_df = county_df[(county_df['year'] >= 2019) & (county_df['year'] <= 2023)]

county_df = county_df[county_df['period'] != 'M13']
county_series_join = pd.merge(county_df, series_df, on= 'series_id                     ')

county_series_join['Employment Measure'] = county_series_join['series_title'].str.split(':').str[0]
county_series_join['County'] = county_series_join['series_title'].str.split(':').str[1].str.split(',').str[0].str.strip()

## CORRECT COUNTY NAMES FOR TABLEAU RECOGNITION

# Replace '/city', '/town', '/municipality', 'Borough', 'Census Area', 'Municipio' with an empty string

county_series_join['County'] = county_series_join['County'].str.replace(r'(/city|/town|/municipality|Borough|Census Area|Municipio)', '', regex=True)

county_series_join['State Abbreviation'] = county_series_join['series_title'].str.split(',').str[1].str.split('(').str[0].str.strip()
county_series_join['State'] = county_series_join['State Abbreviation'].map(state_lookup)
county_series_join.to_csv('county_series_join.csv', index=False)
```
